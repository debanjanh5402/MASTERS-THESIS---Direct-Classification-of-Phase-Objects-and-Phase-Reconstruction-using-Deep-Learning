```python
import os 
import cv2
from tqdm.auto import tqdm
from utility_functions.segmentation_utility import crop_cells


root_dir = "/Users/debanjan_5402/Desktop/Dataset/RBC_Data/defocused"

for fname in tqdm(os.listdir(root_dir), desc="Cropping Cells"):
    if fname.endswith("a.png"):
        holo_path = os.path.join(root_dir, fname)
        bf_path = os.path.join(root_dir, fname.replace("a.png",".png"))

        holo_img = cv2.imread(holo_path)
        bf_image = cv2.imread(bf_path)

        _, _, _, _ = crop_cells(hologram_img=holo_img, brightfield_img=bf_image, 
                                save_dir="./dataset/defocused", filename=fname.replace("a.png",""))

```
